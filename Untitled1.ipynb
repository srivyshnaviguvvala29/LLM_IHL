```python
import gradio as gr

def chatbot_response(message, history):

    message = message.lower()

    if "hello" in message or "hi" in message:
        return "Hello! 👋 I am your Smart Attendance Assistant. How can I help you?"

    elif "attendance" in message:
        return "You can check your overall attendance and subject-wise attendance from the Attendance section."

    elif "percentage" in message:
        return "Your attendance percentage is calculated based on the classes attended and the total classes conducted."

    elif "subject" in message:
        return "You can view attendance details separately for each subject."

    elif "low attendance" in message or "shortage" in message:
        return "If your attendance is below the required percentage, you will receive a low-attendance alert."

    elif "present" in message:
        return "Your present days are recorded when you attend the class and your faculty marks your attendance."

    elif "absent" in message:
        return "Your attendance is marked absent when you do not attend a class."

    elif "faculty" in message or "teacher" in message:
        return "Faculty members can select their subject and class to mark or update student attendance."

    elif "login" in message:
        return "Students, faculty, and administrators can log in using their registered credentials."

    elif "report" in message:
        return "You can view attendance reports including overall, subject-wise, and monthly attendance."

    elif "admin" in message:
        return "The administrator can manage students, faculty, subjects, and attendance reports."

    elif "thank" in message:
        return "You're welcome! 😊"

    elif "bye" in message:
        return "Goodbye! 👋 Have a great day!"

    else:
        return "I'm sorry, I couldn't understand your question. Please ask me about attendance, percentage, subjects, reports, faculty, or login."

demo = gr.ChatInterface(
    fn=chatbot_response,
    title="🤖 Campus Attendance Assistant",
    description="Ask me about attendance problems"
)

demo.launch(share=True)
```

```python
from langchain_community.document_loaders import PyPDFLoader

loader = PyPDFLoader(pdf_file)

documents = loader.load()

print("Number of pages:", len(documents))
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter

text_splitter = RecursiveCharacterTextSplitter(
    chunk_size=500,
    chunk_overlap=50
)

chunks = text_splitter.split_documents(documents)

print("Number of chunks:", len(chunks))
```

```python
from langchain_google_genai import GoogleGenerativeAIEmbeddings

embeddings = GoogleGenerativeAIEmbeddings(
    model="models/gemini-embedding-001"
)

print("Embeddings created successfully!")
```

```python
from langchain_community.vectorstores import FAISS

vectorstore = FAISS.from_documents(
    chunks,
    embeddings
)

print("FAISS vector database created successfully!")
```

```python
retriever = vectorstore.as_retriever(
    search_kwargs={"k": 3}
)

print("Retriever created successfully!")
```

```python
from langchain_google_genai import ChatGoogleGenerativeAI

llm = ChatGoogleGenerativeAI(
    model="gemini-3.8-flash",
    temperature=0
)

print("Gemini 3.8 Flash connected successfully!")
```

```python
def chatbot_response(message, history):

    # 1. Search the attendance document
    relevant_docs = retriever.invoke(message)

    # 2. Get the relevant text
    context = "\n\n".join(
        doc.page_content for doc in relevant_docs
    )

    # 3. Give the retrieved information to Gemini
    prompt = f"""
You are a Smart Attendance Management Assistant.

Answer the user's question using ONLY the information
provided in the context below.

If the answer is not available in the context,
say:
"I could not find this information in the attendance document."

Give a simple and clear answer.

Context:
{context}

User Question:
{message}

Answer:
"""

    # 4. Generate answer
    response = llm.invoke(prompt)

    return response.content
```

```python
import gradio as gr

def chatbot_ui(message, history):
    try:
        response = chatbot_response(message, history)

        # Gemini response ni plain text ga convert cheyyadam
        if isinstance(response, str):
            return response

        if isinstance(response, list):
            texts = []
            for item in response:
                if isinstance(item, dict) and "text" in item:
                    texts.append(item["text"])
            return "\n".join(texts)

        return str(response)

    except Exception as e:
        return f"Error: {str(e)}"


demo = gr.ChatInterface(
    fn=chatbot_ui,
    title="🤖 Campus Attendance Assistant",
    description="Ask me about attendance, percentage, subjects, reports, faculty, and login."
)

demo.launch(share=True)
```
